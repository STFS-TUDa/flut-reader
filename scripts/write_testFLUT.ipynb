```python
import pyFLUT
import collections
import pyFLUT
import numpy as np
```

# Create test table
This script creates a table for testing.

Currently the output is defined as:
- out_1 = x + y
- out_2 = x * y + z
- znew = z + x + y

```python
input_dict = collections.OrderedDict(
    (
        ("x", np.linspace(0, 0.1, 3)),
        ("y", np.linspace(0, 1, 2)),
        ("z", np.linspace(0, 4, 5)),
    )
)
output_list = ["out_1", "out_2"]
data_array = np.ones(
    [len(values) for values in input_dict.values()] + [len(output_list)]
)

flut = pyFLUT.flut.Flut(
    input_dict=input_dict, output_dict=output_list, data=data_array
)
x, y, z = np.meshgrid(*input_dict.values(), indexing="ij")
flut["out_1"] = x + y
flut["out_2"] = x * y + z
flut["znew"] = z + x + y

flut_reader_settings = {"transportedSpecies": {
        "useTransportedSpecies": True,
        "speciesNames": ["CO2", "CO", "CH4", "CH3OCH3"],
        "calculationWeights": {
            "yc": {"CO2": 1.0, "CO": 1.0},
            "ZBilger": {"CO": 0.5, "CO2": 0.075}
        }
    }
}

flut.calc_non_dimensional_quantity(along_variable="x", quantity="x")
flut = flut.map_variables(from_inp="x", to_inp="xnorm")

flut.calc_non_dimensional_quantity(along_variable="z", quantity="znew")
flut = flut.map_variables(from_inp="z", to_inp="znewnorm")

flut.write_h5("testFLUT_norm.h5")
```

# Create input-value pairs inside the table

```python
xs = np.linspace(0, 0.1, 3)
ys = np.linspace(0, 1, 3)
zs = np.linspace(0, 4, 3)

for x in xs:
    for y in ys:
        for z in zs:
            print("{" + f"{x}, {y}, {round(x+y+z, 6)}" + "}")
```
